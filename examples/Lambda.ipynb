```rwlog
theory bound_vars {
    # Unequal variables
    constraint neq/2

    (neq $x $x) <=> fail.
    # Doesn't apply to non-variables
    (neq (lam $v $x) $y) <=> fail.
    (neq $y (lam $v $x)) <=> fail.
    (neq (app $v $x) $y) <=> fail.
    (neq $y (app $v $x)) <=> fail.

    constraint var/1
    (var (app $x $y)) <=> fail.
    (var (lam $x $y)) <=> fail.

    constraint norm/1
    (norm (app (lam $x $y) $z)) <=> fail.
    (norm (app (app $x $y) $z)) <=> (norm (app $x $y)), (norm $z).
    (norm (app $x $z)), (var $x) <=> (var $x), (norm $z).
    (norm (lam $x $y)) <=> (norm $y).
    (norm $x), (var $x) <=> (var $x).
}

rel lamRW {
    # lam[x, x][z] == z
      (app (lam $x $x) $z) -> $z
    # lam[x, y][z] == y if x != y
    | (app (lam $x $y) $z) { (neq $x $y) } -> $y
    # lam[x, lam[x, y]][z] == lam[x, y]
    | (app (lam $x (lam $x $y)) $z) -> (lam $x $y)
    # lam[x, lam[y, z]][w] == lam[y, lam[x, z][w]]
    | (app (lam $x (lam $y $z)) $w) { (neq $x $y) } -> (lam $y (app (lam $x $z) $w))
    # lam[x, y[z]][w] == lam[x, y][w][lam[x, z][w]]
    | (app (lam $x (app $y $z)) $w) -> (app (app (lam $x $y) $w) (app (lam $x $z) $w))
    # Congr Lam
    | [ [(lam $x $y) -> $y ; lamRW ; $y -> (lam $x $y)] &
        (lam $x $y) -> (lam $x $z) ]
    # Congr App L
    | [ [(app $x $y) -> $x ; lamRW ; $x -> (app $x $y)] &
        [(app $x1 $y) -> (app $x2 $y)] ]
    # Congr App R
    | [ [(app $x $y) {(norm $x)} -> $y ; lamRW ; $y -> (app $x $y)] &
        [(app $x $y1) -> (app $x $y2)] ]
}

rel lamEq {
    # base case
      $x {(norm $x)} -> $x
    # rec case
    | lamRW ; lamEq
}
```
